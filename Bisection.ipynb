# Pencarian Akar Persamaan Nonlinear dengan Metode Bisection

Program ini menghitung pendekatan akar persamaan nonlinear dengan membagi interval yang memiliki perubahan tanda secara berulang. Perhitungan dihentikan ketika nilai fungsi mendekati nol atau batas iterasi tercapai.

```python
# Mendefinisikan fungsi nonlinear
def f(x):
    return x**3 - x - 2

# Menentukan interval pencarian akar
a = 1.0
b = 2.0

# Menentukan ketelitian dan batas iterasi
toleransi = 1e-6
maks_iterasi = 100

print("f(a) =", f(a))
print("f(b) =", f(b))
```

digunakan untuk mendefinisikan fungsi nonlinear \(f(x)=x^3-x-2\) yang akan dicari akarnya. Variabel a dan b menentukan batas bawah dan batas atas interval awal, yaitu 1 dan 2. Variabel toleransi digunakan untuk menentukan tingkat ketelitian, sedangkan maks_iterasi membatasi jumlah pengulangan perhitungan. Program juga menampilkan nilai fungsi pada kedua batas interval untuk mengetahui apakah interval memenuhi syarat metode Bisection.

```python
# Menyiapkan tempat menyimpan hasil iterasi
riwayat = []
akar = None

# Memeriksa apakah interval awal memenuhi syarat
if f(a) * f(b) >= 0:
    raise ValueError(
        "Interval salah. f(a) dan f(b) harus berbeda tanda."
    )

# Melakukan iterasi metode Bisection
for iterasi in range(1, maks_iterasi + 1):

    # Menghitung titik tengah
    c = (a + b) / 2
    fc = f(c)

    # Menyimpan hasil iterasi
    riwayat.append([iterasi, a, b, c, fc])

    # Menghentikan perhitungan jika toleransi tercapai
    if abs(fc) < toleransi or (b - a) / 2 < toleransi:
        akar = c
        break

    # Memperbarui interval berdasarkan tanda fungsi
    if f(a) * fc < 0:
        b = c
    else:
        a = c

# Menampilkan hasil pendekatan akar
if akar is not None:
    print("Proses Bisection selesai.")
    print("Jumlah iterasi:", len(riwayat))
    print(f"Perkiraan akar: {akar:.8f}")
    print(f"Nilai f(akar): {f(akar):.8f}")
else:
    print("Batas iterasi tercapai.")
    print(f"Perkiraan akar: {(a + b) / 2:.8f}")
```

merupakan bagian utama program yang melakukan pencarian akar. Program menghitung titik tengah interval menggunakan rumus \(c=(a+b)/2\), kemudian menghitung nilai fungsi pada titik tersebut. Hasil setiap iterasi disimpan dalam variabel riwayat.

Jika nilai fungsi sudah memenuhi toleransi atau interval sudah cukup kecil, proses dihentikan. Jika belum, program memperbarui batas interval berdasarkan tanda fungsi, kemudian mengulangi perhitungan sampai ditemukan perkiraan akar atau jumlah iterasi maksimum tercapai. Hasil akhirnya menampilkan jumlah iterasi, perkiraan akar, dan nilai fungsi pada akar tersebut.

```python
# Menampilkan tabel hasil iterasi metode Bisection
print("Iterasi | a | b | c | f(c)")

# Memastikan riwayat sudah tersedia
if "riwayat" not in globals():
    print("Variabel riwayat belum tersedia.")
    print("Jalankan kembali Cell 3 terlebih dahulu.")
else:
    for data in riwayat:
        i, a_i, b_i, c_i, fc_i = data

        print(
            f"{i:7d} | {a_i:.6f} | {b_i:.6f} | "
            f"{c_i:.6f} | {fc_i:.6f}"
        )
```

digunakan untuk menampilkan riwayat perhitungan metode Bisection dalam bentuk tabel. Program mengambil data yang sudah disimpan dalam variabel riwayat, kemudian menampilkan nomor iterasi, batas bawah (a), batas atas (b), titik tengah (c), dan nilai fungsi pada titik tengah (f(c)).

Tabel ini membantu melihat perubahan interval pada setiap iterasi hingga diperoleh perkiraan akar. Cell 4 harus dijalankan setelah Cell 3 berhasil agar variabel riwayat tersedia.
