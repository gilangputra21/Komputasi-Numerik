```python
# Koefisien x, y, z dan konstanta setiap persamaan
A = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

# Menampilkan matriks awal
print("Matriks augmented awal:")
for baris in A:
    print(baris)
```

setiap baris berisi koefisien \(x,y,z\) dan konstanta di sisi kanan persamaan. Matriks ini menjadi dasar perhitungan selanjutnya.

```python
# Membuat salinan matriks supaya data awal tetap tersedia
M = [baris[:] for baris in A]
n = 3

# Melakukan eliminasi pada setiap kolom pivot
for kolom in range(n):

    # Memilih baris dengan pivot terbesar secara absolut
    pivot_baris = max(
        range(kolom, n),
        key=lambda i: abs(M[i][kolom])
    )

    # Menukar baris jika diperlukan
    M[kolom], M[pivot_baris] = (
        M[pivot_baris], M[kolom]
    )

    pivot = M[kolom][kolom]

    # Jika pivot nol, matriks tidak bisa langsung dieliminasi
    if abs(pivot) < 1e-12:
        raise ValueError(
            "Sistem tidak memiliki solusi tunggal."
        )

    # Menghilangkan nilai di bawah pivot
    for i in range(kolom + 1, n):
        faktor = M[i][kolom] / pivot

        for j in range(kolom, n + 1):
            M[i][j] -= faktor * M[kolom][j]

    print(f"\nMatriks setelah kolom {kolom + 1}:")
    for baris in M:
        print([round(v, 4) for v in baris])
```

program memilih pivot yang sesuai, menukar baris bila perlu, lalu menghilangkan koefisien di bawah pivot. Pemilihan pivot terbesar secara absolut membantu mengurangi masalah pembulatan numerik.

```python
# Menyiapkan tempat untuk menyimpan x, y, dan z
solusi = [0.0] * n

# Menghitung variabel dari baris terakhir ke pertama
for i in range(n - 1, -1, -1):
    jumlah = sum(
        M[i][j] * solusi[j]
        for j in range(i + 1, n)
    )

    solusi[i] = (
        M[i][n] - jumlah
    ) / M[i][i]

# Menampilkan hasil akhir
print("Hasil penyelesaian SPL:")
print(f"x = {solusi[0]:.4f}")
print(f"y = {solusi[1]:.4f}")
print(f"z = {solusi[2]:.4f}")
```

nilai \(z\) dihitung terlebih dahulu, kemudian \(y\), dan terakhir \(x\). Hasil akhirnya ditampilkan hingga empat angka di belakang koma.

```python
# Menguji hasil dengan memasukkan solusi ke persamaan awal
for i, baris in enumerate(A, start=1):
    kiri = sum(
        baris[j] * solusi[j]
        for j in range(n)
    )
    kanan = baris[n]

    print(
        f"Persamaan {i}: "
        f"hasil = {kiri:.4f}, "
        f"konstanta = {kanan}"
    )
```

hasil setiap persamaan dibandingkan dengan konstantanya. Jika hasilnya sama, solusi memenuhi persamaan awal, dengan mempertimbangkan pembulatan angka desimal.
