```python
# Koefisien persamaan x, y, z dan konstanta
A = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

print("Matriks awal:")
for baris in A:
    print(baris)
```

Cell 1 digunakan untuk memasukkan koefisien dari tiga persamaan linear ke dalam bentuk matriks augmented. Variabel A menyimpan koefisien \(x\), \(y\), \(z\), serta konstanta di sebelah kanan persamaan. Perulangan for digunakan untuk menampilkan matriks awal sebelum dilakukan proses eliminasi.

```python
# Menyalin matriks untuk proses perhitungan
M = [baris[:] for baris in A]
n = 3

# Eliminasi setiap kolom
for k in range(n):
    # Memilih baris dengan pivot terbesar
    p = max(range(k, n), key=lambda i: abs(M[i][k]))

    # Menukar baris jika diperlukan
    M[k], M[p] = M[p], M[k]

    pivot = M[k][k]

    # Memastikan pivot tidak nol
    if abs(pivot) < 1e-12:
        raise ValueError("Sistem tidak memiliki solusi tunggal")

    # Menghilangkan koefisien di bawah pivot
    for i in range(k + 1, n):
        faktor = M[i][k] / pivot

        for j in range(k, n + 1):
            M[i][j] -= faktor * M[k][j]

print("Matriks setelah eliminasi:")
for baris in M:
    print([round(v, 4) for v in baris])
```

Cell 2 digunakan untuk mengubah matriks awal menjadi matriks segitiga atas. Pertama, matriks A disalin ke variabel M agar data awal tidak berubah. Variabel n menyatakan jumlah variabel, yaitu tiga.

Selanjutnya, program memilih elemen pivot terbesar berdasarkan nilai absolutnya. Jika diperlukan, baris ditukar agar perhitungan lebih stabil. Program kemudian menggunakan operasi baris elementer untuk menghilangkan koefisien di bawah pivot pada setiap kolom.

Setelah seluruh proses selesai, program menampilkan matriks hasil eliminasi yang akan digunakan untuk menghitung nilai variabel.

```python
# Menyiapkan tempat penyimpanan solusi
solusi = [0.0] * n

# Menghitung variabel dari bawah ke atas
for i in range(n - 1, -1, -1):
    jumlah = sum(
        M[i][j] * solusi[j]
        for j in range(i + 1, n)
    )

    solusi[i] = (M[i][n] - jumlah) / M[i][i]

print("Perhitungan selesai.")
```

Cell 3 digunakan untuk mencari nilai \(x\), \(y\), dan \(z\) berdasarkan matriks segitiga atas yang diperoleh dari Cell 2. Variabel solusi digunakan untuk menyimpan hasil perhitungan ketiga variabel.

Perulangan dimulai dari baris terakhir menuju baris pertama. Nilai \(z\) dihitung terlebih dahulu, dilanjutkan dengan \(y\), kemudian \(x\). Hasil perhitungan disimpan ke dalam variabel solusi agar dapat ditampilkan pada cell berikutnya.

```python
# Menampilkan hasil penyelesaian SPL
print("Hasil Penyelesaian SPL 3 Variabel")
print("---------------------------------")
print(f"Nilai x = {solusi[0]:.4f}")
print(f"Nilai y = {solusi[1]:.4f}")
print(f"Nilai z = {solusi[2]:.4f}")
```

Cell 4 digunakan untuk menampilkan hasil penyelesaian sistem persamaan linear tiga variabel. Program mengambil nilai dari variabel solusi, kemudian menampilkan nilai \(x\), \(y\), dan \(z\) dengan empat angka di belakang koma menggunakan format :.4f.
